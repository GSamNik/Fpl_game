# copy()

Returns a shallow copy of the list.

What Does "Shallow Copy" Mean?

A shallow copy creates a new list object

```python
l= ['a', 'b' , 'r', 'g', 'r']
help(l.copy)
```

```python
l=[3,7,4,5,7]
z=l.copy() # shallow copy
z
```

```python
z.append(20)
print(z)
print(l)
```

```python
l=[3,7,4,5,7]
# not a real copy —> just a reference
z=l                # both z and l refer to same object so if one changed another change
z.append(10)
print(l)
print(z)
```
