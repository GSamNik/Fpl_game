# Remove a letter (e) from this tuple

t=('a','b','c','d','a','e','a','a','e','e')


```python
t=('a','b','c','d','a','e','a','a','e','e')
list_1=list(t)

for i in range(list_1.count('e')):
  list_1.remove('e')

list_1
```

```python
t=('a','b','c','d','a','a','a','e','e')
list_1=list(t)
i=0
while i < len(list_1):
  if list_1[i]=='e':
    list_1.remove('e')
    continue
  i+=1

list_1
```

```python
t=('a','b','c','d','a','e','a','a','e','e')
l=list(t)
i=0
while i<len(l):
  if l[i]=='e':
     l.remove(l[i])
     continue
  i+=1
t=tuple(l)
t
```

```python
t=('a','b','c','d','a','e','a','a','e','e')
l=list(t)
# l.count('a')
for i in range(l.count('e')):
    l.remove('e')
t=tuple(l)
print(t)
```

```python
t=('a','b','c','d','a','e','a','a','e','e')
l=[]
for i in t:
  if i != 'e':
    l.append(i)
print(l)
```
