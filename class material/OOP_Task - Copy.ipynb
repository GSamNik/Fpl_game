# OOP Task — YouTube Video Management System

```

You are working as a Python developer for a small company that wants to build a simple system to manage YouTube videos.

The company wants every video to have its own information, but they do not want users to directly modify the video's data. For example, a user should not be able to change the number of views to -500 or make the number of likes greater than the number of views.

Your job is to create a YouTubeVideo class that protects its internal data and allows users to interact with it through controlled methods.

Scenario

When a new video is uploaded, the system stores:

The video's title
The channel name
Number of views
Number of likes
Number of subscribers

For example, a video might be:

"Python OOP Tutorial" uploaded by "Rawan AI", with 10,000 views, 800 likes, and 5,000 subscribers.

The company wants this information to be private.

When someone wants to see the information, they should use appropriate getter methods.

When someone wants to update information, they should use setter methods, and the system must validate the new value before accepting it.


Business Rules

The system must follow these rules:

A video title cannot be empty.
A channel name cannot be empty.
Views cannot be negative.
Likes cannot be negative.
Likes cannot be greater than views.
Subscribers cannot be negative.
When a video receives new views, the views should increase.
When someone likes the video, the number of likes increases by 1.

For example, if a video has:

Views: 10,000
Likes: 800

and someone tries to change the likes to 15,000, the system should reject the change because:

Likes cannot be greater than views.



Your Task
Create a class called:

YouTubeVideo

Use Encapsulation to protect the video's data.

Your class should provide a way to:

Create a new video.
View the video's information.
Update the video's information safely.
Add new views.
Like the video.
Calculate the video's like percentage.

Example Scenario
The company uploads a new video:

Title: Python OOP Tutorial
Channel: Rawan AI
Views: 10,000
Likes: 800
Subscribers: 5,000

Later:

1,500 people watch the video.
2 people like the video.
The channel gains 500 subscribers.

At the end, the system should show the updated information.

# Test your system
```
video = YouTubeVideo(
    "Python OOP Tutorial",
    "Rawan AI",
    10000,
    800,
    5000
)

video.display_info()

# People watch the video
video.add_views(1500)

# People like the video
video.like_video()
video.like_video()

# New subscribers
video.add_subscribers(500)

video.display_info()

# Scenario: Search System for an E-Commerce Website

You are developing a search system for an online shopping website.

When a customer searches for a product, the website saves the search in their recent search history so they can easily find it again later.

```
For example, Rawan searches for:

"laptop"
"wireless mouse"
"keyboard"
```
The website displays these searches under Recent Searches.

```
However, the system has some rules:

A search cannot be empty.
Spaces before/after a search should be removed.
The same search should not appear twice.
Each user has a maximum number of recent searches.
If the history is full, the oldest search is removed.
A user can remove a specific search.
A user can clear all searches.
The search history must be protected from direct modification.
```


Your Task

Create a class: SearchSystem

The class should store:

Username
Search history
Maximum number of searches

Use Encapsulation:

private attributes
       ↓
getters / setters
       ↓
validation

Do not allow direct access to the search history.


```
Test Scenario

Imagine that Rawan is using the shopping website.

She is allowed to keep only 3 recent searches.

Step 1 — Create the user
search = SearchSystem("Rawan", 3)

Expected:

User: Rawan
Recent Searches: []
Maximum Searches: 3
Step 2 — Rawan searches for a laptop
search.add_search("laptop")

Expected:

Recent Searches:
1. laptop
Step 3 — She searches for a mouse
search.add_search("wireless mouse")

Expected:

Recent Searches:
1. laptop
2. wireless mouse
Step 4 — She searches for a keyboard
search.add_search("gaming keyboard")

Expected:

Recent Searches:
1. laptop
2. wireless mouse
3. gaming keyboard
Step 5 — She searches for a new product
search.add_search("headphones")

The history is already full.

So the oldest search (laptop) should be removed.

Expected:

Recent Searches:
1. wireless mouse
2. gaming keyboard
3. headphones
Step 6 — She searches for the keyboard again
search.add_search("gaming keyboard")

The system should not create a duplicate.

Instead, it should move the existing search to the most recent position.

Expected:

Recent Searches:
1. wireless mouse
2. headphones
3. gaming keyboard
Step 7 — Remove a search

Rawan doesn't want the mouse search anymore:

search.remove_search("wireless mouse")

Expected:

Recent Searches:
1. headphones
2. gaming keyboard
Step 8 — Invalid search

Rawan accidentally searches with an empty string:

search.add_search("   ")

Expected:

Search cannot be empty.

The history should remain unchanged.

Step 9 — Clear history

At the end, Rawan wants to remove all recent searches:

search.clear_history()

Expected:

Recent Searches: []


- Final Test

Your program should run something similar to:

search = SearchSystem("Rawan", 3)


search.display_history()


search.add_search("laptop")
search.add_search("wireless mouse")
search.add_search("gaming keyboard")


search.display_history()


search.add_search("headphones")


search.display_history()


search.add_search("gaming keyboard")


search.display_history()


search.remove_search("wireless mouse")


search.display_history()


search.add_search("   ")


search.clear_history()


search.display_history()

Important

Don't do this:

search.__search_history.append("laptop")

The whole point of the task is that the SearchSystem controls its own data through methods such as:

get_search_history()
set_search_history()
add_search()
remove_search()
clear_history()

This gives the student a realistic reason to use Encapsulation, instead of simply making private variables because the exercise says so.
