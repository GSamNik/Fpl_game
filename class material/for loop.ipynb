## Section 2: The `in` Operator

*The `in` operator performs a search through a sequence to check whether a value exists. It works similarly to a linear search by checking elements one by one until the value is found or the sequence ends.*

```python
x=range(0,8)
7 in x
```

```python
x=range(0,8)
7 not in x
```

## Section 3: The For Loop
*3.1 A `for` loop is used for iterating over a sequence or iteratble  (that is either a list, a tuple, a dictionary, a set, or a string).*

```python
for i in range(5):
  print(i)
```

```python
for i in range(5):
  print('rawan')
```

```python
l=[1,211,'rawan',True]
for i in l:
  print(i)
```

```python
l=[1,211,'rawan',True]
l[i]
```

```python
l=[1,211,'rawan',True]
for i in range(len(l)):
  print(l[i])
```

```python
l=[1,2,3,4]
for i in l:
  i=i**2
  print(i)

print(l)
```

```python
l=[1,2,3,4]
for i in range(len(l)):
  l[i]=l[i]**2
  print(l[i])

print(l)
```
