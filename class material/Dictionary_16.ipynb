# -- Dictionary in Python --

# [1] Dictionary items are enclosed in curly braces { }
     - A dictionary is created using `{}`.
     - Example: my_dict = {"name": "Rawan", "age": 22}

# [2] Dictionary items are key-value pairs separated by a colon (:)
     - Each item has a key and a value.
     - Example: "name": "Rawan" → "name" is the key, "Rawan" is the value.

# [3] Dictionary keys must be immutable types
    - Immutable types include numbers, strings, and tuples.
    - You cannot use lists or other dictionaries as keys because they are mutable.
#     - Example: {"a": 1} ✅ but {[1, 2]: 3} ❌

# [4] Dictionary values can be of any data type
     - Values can be strings, numbers, lists, dictionaries, etc.
     - Example: {"name": "Rawan", "skills": ["Python", "ML", "DL"]}

# [5] Dictionary keys must be unique
     - You cannot have duplicate keys.
     - If a duplicate key is added, the last value will overwrite the previous one.
#     - Example: {"a": 1, "a": 2} → the value of "a" will be 2

# [6] Dictionaries are unordered (Unindexed)
     - You access dictionary elements using keys, not positions.
     - Example:
       user = {"name": "Rawan", "age": 22}
       print(user["name"]) → outputs "Rawan"

#Pop item

# Copy

creat a shallow copy

```python
user={"name":"rawan","country":"Egypt","skills":["python","ml","Dl"]}
user2=user.copy()   # shallow copy
user2["language"]=["arbic",'en','fr']

print(user)
print(user2)
```

```python
user={"name":"rawan","country":"Egypt","skills":["python","ml","Dl"]}
user2=user  # add new refrence to dict in memory
user2["language"]="arbic"

print(user)
print(user2)
```
