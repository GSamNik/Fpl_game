````python
x=10
def num_add_condtant():
  x+=1 # x=x+1=11
  return x    # cannot be apply modification inside function
print(num_add_condtant())
print(x)
````

````python
x=10
def num_add_condtant():
  global x
  x+=1
  return x    # can be apply operation inside func
print(num_add_condtant())
print(x)
````

| Feature              | `map()`                                     | `filter()`                                                      |
| -------------------- | ------------------------------------------- | --------------------------------------------------------------- |
| **Purpose**          | Transforms each item (modifies or computes) | Selects items that meet a condition (filters)                   |
| **Returns**          | A new iterable with the **modified items**  | A new iterable with the **original items** that passed the test |
| **Function Type**    | Should return the **new value**             | Should return **True or False**                                 |
| **Keeps All Items?** | Yes, all are transformed                    | No, only items where function returns `True`                    |
| **Use When**         | You want to change each item                | You want to filter out items                                    |

#### The map() Function in Python
The map() function applies a given function to all items in an input list (or any other iterable) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively.

````python
def check_even(l):
   n=[]
   for i in l:
      n.append(i % 2==0)
   return n
l=[1,2,3,4,5]
check_even(l)
````
