````python
def add(n1,n2):
  if isinstance(n1,(float,int)):
    if isinstance(n2,(float,int)):
       add=n1+n2
       print(add)
    else:
      print('Second number must be numbers')
      print('invalid data type')
  else:
    print('First number must be numbers')
    print('invalid data type')
add(1,2)
````
