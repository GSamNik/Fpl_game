# List

List → Sequence Data Type

1. Items in square bracket
   -  Example: [1, 2, 3]

2. Has order → zero-indexed

   * First element index is 0.

3. Mutable → changeable
   * You can modify, add, or remove elements.

4. Different data types
   * Elements can be of different types (int, string, etc.)

5. Elements do not have to be unique
   * Duplicates are allowed.

| Feature          | list                                | array                       |
|------------------|--------------------------------------|-----------------------------|
| Data types       | Can mix types                        | Must be same type           |
| Memory efficient | ✖ Not optimized (dynamic)            | ✔ More efficient (static)   |
| Flexible         | More flexible (dynamic and in size)  | Not very flexible           |

# Slice List

## [start:end:step]

```python
#  0  1  2  3  4   5
l=[10,0,33,65,190,900]
# -6 -5 -4 -3  -2  -1
l[0:5] # end exclude end-1
```

```python
l[1:5:1] # [Start(include):End(exclude==end-1):step(defult=1)]
```

```python
#  0  1  2  3  4   5
l=[10,0,33,65,190,900]
# -6 -5 -4 -3  -2  -1
l[0:6:2]
```

```python
l[1:]      # 1: --> means to end
```

```python
l[:4]    # :4 --> from frist element
```

```python
l[:]   # : --> from frist element to the last
```

```python
l[::2] # even index
```

```python
#  0  1  2  3  4   5
l=[10,0,33,65,190,900]
# -6 -5 -4 -3  -2  -1
l[-4:-1]         # -4 --> -1-1=-2 step is defult = 1 means go forward
```

```python
l[-5:]
```

```python
l=[10,0,33,65,190,900]
# -6 -5 -4 -3  -2  -1
l[-1:-4]
```

```python
l=[10,0,33,65,190,900]
# -6 -5 -4 -3  -2  -1
l[5:2]
```

```python
l[-1:-4:-1]
```

```python
l[::-1] # backward # reverse list
```

# Golden Rule

### Element Assignment

Replaces one specific element at the given index.

furits= ['apple', 'mango', 'cherry', 'kiwi', 'guava']

- fruits[1] = "watermelon"

- Output:

           ['apple', 'watermelon', 'cherry', 'kiwi', 'guava']

### Slice Assignment

Replaces a portion of the list (a slice), not just a single element.
The value on the right-hand side must be an iterable (such as a list, tuple, or string).

- fruits[1:] = ["watermelon"]

- Output:

        ['apple', 'watermelon']

```python
l=[10,0,33,65,190,900]
l[1:]=111
```

```python
l=[10,0,33,65,190,900]
l[1:]=[111,222,333]
l
```

```python
l=[10,0,33,65,190,900]
l[1:3]='rawan'
l
```
