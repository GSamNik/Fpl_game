# Dealing with data

- Applications are built from **code and data**.
- **Code** is what you write to **handle and control (handle it) data**.
- To make sense of data, we need to **organize it into types**, like:
  - **Numbers** (`int`, `float`)
  - **Strings** (`str`)
  - **Booleans** (`True`, `False`)
- Each type is used for different purposes:
  - 📊 **Numbers** →  age, salaries
  - 🔤 **Strings** →  names, position
  - ✅ **Booleans** → whether a instrcutor got paid True or False.
- This data is **stored in memory**, and we use **variables to refer to it**.
- A variable doesn’t store im memory — it **points or refer to the location in memory**.
- Code use data to preform operations

---

## ❓ Why Do Data Types Exist?

Because computers need to:

1. **Know how much memory to use** for each type (e.g., an `int` vs a `str`)
2. **Understand what operations** are allowed:
   - You can **add numbers**
   - You can **concatenate strings**
   - ❌ But adding a number to a string directly causes an error

---

#**Dynamic VS Static Languages**
## Statically Typed Languages (e.g., Rust, C, C++)

- Variable types are **static**; once declared, you **cannot change** them.
- The type of a variable is known at **compile time** ("type checking at compile time").
- Have **better performance** at runtime due to **not needing to check types dynamically**.
- ✅ **Errors are detected earlier** (before execution).

---

## Dynamically Typed Languages (e.g., PHP, Python, JavaScript)

- Variable types are **dynamic**; you can always **change** them.
- **Type checking happens at execution time**.
- ⚠️ **Errors can be detected after execution**.


## The `type()` Function
