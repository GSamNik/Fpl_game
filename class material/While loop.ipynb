`while` loops run as long as a condition is `True`. They are useful when we don't know beforehand how many times the loop needs to run.*

# While Loop Syntax
```
# Initialization

while condition:
     Block of code
     Increment / Update
```

````python
min_input=int(input(' plz min '))
max_input=int(input(' plz max '))

if min_input <= max_input:
  while min_input <= max_input:
    print(min_input)
    # i=i+1
    min_input+=1
else:
  print('min greater than max')
````

### 1.1 Basic Syntax
*A simple counter using a `while` loop.*

````python
i = 0

while i <= 10:
    print("Current count: ", i)
    i += 1  # Increment to prevent infinite loop
````

### 1.2 Handling Negative Ranges
*Loops can handle negative starting points and classify data.*

````python
i = -5

while i <= 5:

    if i < 0:
        print("Negative Number:",i)
    elif i==0:
        print('Zero',i)
    else:
        print("Positive Number:",i)
    i += 1
````

## Section 2: Loop Control (`break` and `continue`)
*Controlling the flow of the loop manually.*

````python
i = 0

while i <= 10:
    if i == 5:
        print("Breaking loop at 5")
        break # Exit loop immediately
    print(i)
    i += 1
````

````python
i = 0

while i <= 10:
    if i == 5:
        print("Breaking loop at 5")
        break # Exit loop immediately
        print('Loop breaked')
    print(i)
    i += 1
print('Finish while')
````

### 2.2 The `continue` Statement
*Skips the rest of the code in the current iteration and jumps to the next one.*

````python
i=0

while i <=5:
  if i==3:
    i+=1
    continue
  print(i)
  i+=1

````

````python
i = 0

while i <= 10:
    i += 1
    if i == 5:
        continue # Skip printing 5
    print(i)
````

````python
i=0
while i <=10:
  if i==5:
    i+=1
    continue # Skip printing 5
  print(i)
  i+=1
````

````python
i=0
while i <=10:
  if i%2==0:
    i+=1
    continue # Skip even number
  print(i)
  i+=1
````
